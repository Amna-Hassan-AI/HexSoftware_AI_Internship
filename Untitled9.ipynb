```python
def check_emotion(text):
    analysis = TextBlob(text)
    # Polarity check karta hai: 1 ka matlab bohat khush, -1 ka matlab bohat dukhi
    score = analysis.sentiment.polarity

    if score > 0.5:
        return "Bohat Khush! 😊"
    elif score > 0:
        return "Theek Thaak / Positive 🙂"
    elif score == 0:
        return "Neutral / Koi jazbaat nahi 😐"
    else:
        return "Thoda Pareshan ya Sad 😟"

# Test karne ke liye yahan kuch likhen
user_input = "I am so excited to start this AI project!"
print(f"Emotion: {check_emotion(user_input)}")
```

```python
!pip install googletrans==4.0.0-rc1
from googletrans import Translator

def smart_emotion_detector():
    translator = Translator()
    print("--- 🧠 Nexora Smart Emotion AI ---")
    text = input("Kuch likhen (Urdu ya English): ")

    # Urdu ko English mein translate karna taake model samajh sake
    translated = translator.translate(text, dest='en').text

    # Emotion check karna
    analysis = TextBlob(translated)
    score = analysis.sentiment.polarity

    print(f"\nTranslated (for AI): {translated}")

    if score > 0.5:
        print("Result: Bohat Zyada Khushi! 🌟😍")
    elif score > 0:
        print("Result: Positive / Acha jazba 🙂")
    elif score == 0:
        print("Result: Neutral / Normal 😐")
    elif score > -0.5:
        print("Result: Thoda Sad ya Negative 😟")
    else:
        print("Result: Bohat Ghussa ya Gham 😡💔")

# Run karne ke liye
smart_emotion_detector()
```
